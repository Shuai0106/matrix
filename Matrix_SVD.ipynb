```python
import numpy as np
```

```python
np.random.seed(42)
```

```python
matrix = np.random.rand(3,3)
```

```python
print('原始矩阵:\n',matrix)
```

```python
U,Sigma,VT = np.linalg.svd(matrix)
```

```python
print('U矩阵:\n',U)
```

```python
print('奇异值:\n',Sigma)
```

```python
print('VT矩阵:\n',VT)
```

```python
Sigma_mat = np.diag(Sigma)
```

```python
reconstructed_mat = U@Sigma_mat@VT
```

```python
print('重构矩阵:\n',reconstructed_mat)
```
